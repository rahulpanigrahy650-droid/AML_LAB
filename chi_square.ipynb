```python
import pandas as pd


rows = int(input("enter number of rows :"))
cols = int(input("enter number of columns :"))

columns =[]
for i in range(cols):
    name  = input(f"enter column {i+1} name :")
    columns.append(name)


data = []

for i in range(rows):
    row =[]
    for j in range(cols):
        value = int(input(f"enter value for {columns[j]} in row {i+1}"))
        row.append(value)
    data.append(row)

df = pd.DataFrame(data,columns  = columns,index =["yes","no"])
print(df)

#row total

row_total = df.sum(axis=1)

#cols total

cols_total = df.sum(axis=0)

#grand total

total = df.values.sum()

#expectedvalues

expected = pd.DataFrame(index = df.index,columns = df.columns)


for i in range(rows):
    for j in range(cols):
        expected.iloc[i,j] = (row_total.iloc[i]* cols_total.iloc[j]/total)



print("\n Expected Data :")
print(expected)

#chi square calculattion

chi_sqaure = ((df - expected)**2/expected).sum().sum()

print("\n calculated chi-sqaure =",chi_sqaure)

#degree of freedom

dof = (rows -1)  * (cols -1)

print("\n Degreeof FReedom :",dof)

#tabulate chi_sqaure table

tabulated_value =float(input("enter tabulated value :"))

print("\n Tabulated value :",tabulated_value)

#hypothesis decission

if chi_sqaure <= tabulated_value :
    print("Null Hypothesis is accepted")
else:
    print("null hypothesuis rejected")
```
