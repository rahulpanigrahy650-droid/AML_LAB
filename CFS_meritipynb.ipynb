```python
import pandas as pd


rows = int(input("enter no of rows :"))
cols = int(input("enter no of cols :"))


columns = []
for i in range(cols):
    columns.append(input("enter column name :"))


data = []
for i in range(rows):
    row = []
    for j in range(cols):
        row.append(input("enter value :"))
    data.append(row)

df = pd.DataFrame(data, columns=columns)
print(df)
```

```python
import numpy as np
import pandas as pd


df = df.apply(pd.to_numeric, errors="coerce")


rx1_y = abs(df["x1"].corr(df["y"]))
rx2_y = abs(df["x2"].corr(df["y"]))
rx3_y = abs(df["x3"].corr(df["y"]))

rx1_x2 = abs(df["x1"].corr(df["x2"]))
rx2_x3 = abs(df["x2"].corr(df["x3"]))
rx3_x1 = abs(df["x3"].corr(df["x1"]))


subsets = [
    (["x1"], 1, [rx1_y], [0]),
    (["x2"], 1, [rx2_y], [0]),
    (["x3"], 1, [rx3_y], [0]),
    (["x1", "x2"], 2, [rx1_y, rx2_y], [rx1_x2]),
    (["x2", "x3"], 2, [rx2_y, rx3_y], [rx2_x3]),
    (["x1", "x3"], 2, [rx1_y, rx3_y], [rx3_x1]),
    (["x1", "x2", "x3"], 3, [rx1_y, rx2_y, rx3_y], [rx1_x2, rx2_x3, rx3_x1]),
]


best_features = None
highest_merit = -1

print("CFS MERIT RESULTS using pipeline  :")
for features, k, r_cf_list, r_ff_list in subsets:

    mean_rcf = np.mean(r_cf_list)
    mean_rff = np.mean(r_ff_list)

    numerator = k * mean_rcf
    denominator = np.sqrt(k + (k * (k - 1) * mean_rff))
    merit_score = numerator / denominator

    print(f"Features: {features} = Merit: {merit_score:.4f}")

    if merit_score > highest_merit:
        highest_merit = merit_score
        best_features = features

print("\n BEST RESULT  : ")
print(f"Optimal Subset: {best_features}")
print(f"Highest Score: {highest_merit:.4f}")
```
